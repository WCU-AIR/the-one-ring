```python
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName("demo").getOrCreate()
sc = spark.sparkContext          # RDD / low-level API entry point
sc.defaultParallelism            # e.g. total cores across live workers
```

```python
textFile = sc.textFile("/opt/spark-data/gutenberg-shakespeare.txt")
```

```python
%%time
textFile.count()
```

```python
wordcount = textFile.flatMap(lambda line: line.split(" ")) \
            .map(lambda word: (word, 1)) \
            .reduceByKey(lambda a, b: a + b)
```

```python
wordcount.saveAsTextFile("/opt/spark-apps/output/wc_output_01")
```
